```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('data/train.txt',sep=';',header=None,names=['text','emotion'])
```

```python
df
```

```python
df.isnull().sum()
```

```python
unique_emotions=df['emotion'].unique()
emotion_nos={}
i=0
for e in unique_emotions:
    emotion_nos[e]=i
    i+=1
df['emotion']=df['emotion'].map(emotion_nos)
df
```

```python
df['text']=[str(x).lower() for x in df['text']]
df
```

```python
import string
def remove_punc(text):
    return text.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
def remove_nos(text):
    new=''
    for i in text:
        if not i.isdigit():
            new+=i
    return new
df['text']=df['text'].apply(remove_nos)
```

```python
def remove_emojis(text):
    new=''
    for i in text:
        if i.isascii():
            new+=i
    return new
df['text']=df['text'].apply(remove_emojis)
```

```python
stop_words=set(stopwords.words('english'))
stop_words
```

```python
def remove_stop_words(text):
    words=text.split()
    cleaned=[]
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
    return ' '.join(cleaned)
```

```python
df['text']=df['text'].apply(remove_stop_words)
```
